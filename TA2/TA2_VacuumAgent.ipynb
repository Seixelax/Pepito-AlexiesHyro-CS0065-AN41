```python
import random

class VacuumAgent:
    def __init__(self, starting_room="A"):
        # init starting pos
        self.location = starting_room

    def perceive_and_act(self, env):
        # check state and execute corresponding rule
        if env[self.location] == "Dirty":
            print(f"Action: Cleaned Room {self.location}")
            env[self.location] = "Clean"
        else:
            # bonus task: find dirty rooms and randomly pick one to target
            dirty_rooms = [room for room, state in env.items() if state == "Dirty"]

            if dirty_rooms:
                next_room = random.choice(dirty_rooms)
                print(f"Action: Room {self.location} is Clean. Moving to Room {next_room}.")
                self.location = next_room
            else:
                print(f"Action: All rooms are clean. Idling in Room {self.location}.")

    def render_grid(self, env):
        # bonus task: text-based grid visualization
        # D = Dirty, C = Clean, * = Agent Location
        grid = ""
        for room in ["A", "B", "C"]:
            agent_mark = "*" if self.location == room else " "
            state_mark = "D" if env[room] == "Dirty" else "C"
            grid += f"[{room}:{state_mark} {agent_mark}]  "
        print(f"Map: {grid}")

def setup_environment():
    # prompt user config for initial room states (now includes C)
    env = {}
    print("Set room states ('Dirty' or 'Clean'):")
    for room in ["A", "B", "C"]:
        env[room] = input(f"Room {room} state: ").strip().capitalize()
    return env

def run_simulation():
    env = setup_environment()

    try:
        steps = int(input("Enter simulation step count: "))
    except ValueError:
        print("Invalid step input. Defaulting to 5 steps.")
        steps = 5

    agent = VacuumAgent(starting_room="A")

    # execute simulation loop
    for step in range(steps):
        print(f"\n--- Step {step + 1} ---")
        agent.render_grid(env)
        agent.perceive_and_act(env)

    print("\n--- Final Status ---")
    agent.render_grid(env)

# trigger run
if __name__ == "__main__":
    run_simulation()
```
