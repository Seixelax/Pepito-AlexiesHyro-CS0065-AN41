```python
from textblob import TextBlob

# run sentiment analysis across 10 input sentences
counts = {"Positive": 0, "Neutral": 0, "Negative": 0}
num_runs = 10

for i in range(1, num_runs + 1):
    sentence = input(f"[{i}/{num_runs}] Enter a sentence: ")
    s = TextBlob(sentence)
    p = s.sentiment.polarity

    if p < 0:
        label = "Negative"
        counts["Negative"] += 1
    elif p == 0:
        label = "Neutral"
        counts["Neutral"] += 1
    else:
        label = "Positive"
        counts["Positive"] += 1

    print(f"Polarity: {p:.2f} | Subjectivity: {s.sentiment.subjectivity:.2f} -> {label}\n")

print("--- Summary ---")
print(f"Positive: {counts['Positive']}")
print(f"Neutral:  {counts['Neutral']}")
print(f"Negative: {counts['Negative']}")
```

```python
import re

# lexicon sets (bing/afinn-inspired baseline word lists)
HAPPY_WORDS = {
    "happy", "joy", "joyful", "love", "loved", "loving", "great", "good",
    "awesome", "amazing", "fantastic", "wonderful", "excited", "glad",
    "pleased", "delighted", "cheerful", "optimistic", "hopeful", "success",
    "win", "winning", "calm", "peaceful", "grateful", "thankful", "brilliant"
}

SAD_WORDS = {
    "sad", "unhappy", "sorrow", "sorrowful", "depressed", "depression",
    "miserable", "hate", "hated", "bad", "terrible", "awful", "horrible",
    "upset", "angry", "mad", "anxious", "worried", "afraid", "scared",
    "lonely", "alone", "cry", "crying", "tears", "fail", "failed", "loss", "lost"
}

def _tokenize(sentence):
    # extract lowercase alphabetic tokens
    return re.findall(r"[a-z']+", sentence.lower())

def number_happy(sentence):
    tokens = _tokenize(sentence)
    return sum(1 for word in tokens if word in HAPPY_WORDS)

def number_sad(sentence):
    tokens = _tokenize(sentence)
    return sum(1 for word in tokens if word in SAD_WORDS)

def analyze_tone(sentence):
    h_count = number_happy(sentence)
    s_count = number_sad(sentence)

    if h_count > s_count:
        tone = "happy"
        sentiment = "positive"
    elif s_count > h_count:
        tone = "sad"
        sentiment = "negative"
    else:
        tone = "neutral"
        sentiment = "neutral"

    print(f"Happy words: {h_count}")
    print(f"Sad words: {s_count}")
    print(f"Tone: {tone} ({sentiment} sentiment)")

if __name__ == "__main__":
    user_sentence = input("Enter a sentence: ")
    analyze_tone(user_sentence)
```
